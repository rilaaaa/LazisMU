```python
import sqlite3
import subprocess

def create_sqlite_dump(db_path, output_file):
    """Membuat dump SQL dari database SQLite"""
    try:
        # Menggunakan sqlite3 command line tool untuk membuat dump
        command = f'sqlite3 {db_path} .dump > {output_file}'
        subprocess.run(command, shell=True, check=True)
        print(f"Backup SQLite berhasil dibuat di: {output_file}")
    except subprocess.CalledProcessError as e:
        print(f"Gagal membuat backup: {e}")

# Contoh penggunaan
sqlite_db = '/Users/zulfanfaizun/Documents/ZULFAN/DEV/LazisMU/test/db.sqlite3'
backup_file = 'sqlite_backup.sql'
create_sqlite_dump(sqlite_db, backup_file)
```
